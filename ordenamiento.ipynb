```python
import timeit
import random

# Implementación del algoritmo de ordenamiento Burbuja
def metodo_burbuja(lista):
    for i in range(len(lista)):
        for j in range(0, len(lista) - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
    return lista

# Implementación del algoritmo de ordenamiento Quicksort
def quicksort(lista):
    if len(lista) <= 1:
        return lista
    else:
        pivote = lista[len(lista) // 2]
        izquierda = [x for x in lista if x < pivote]
        centro = [x for x in lista if x == pivote]
        derecha = [x for x in lista if x > pivote]
        return quicksort(izquierda) + centro + quicksort(derecha)

# Generar listas de diferentes tamaños
listas = {
    100: [random.randint(0, 10000) for _ in range(100)],
    1000: [random.randint(0, 10000) for _ in range(1000)],
    5000: [random.randint(0, 10000) for _ in range(5000)],
    7500: [random.randint(0, 10000) for _ in range(7500)],
    10000: [random.randint(0, 10000) for _ in range(10000)]
}

# Pruebas de rendimiento
print("Comparación de rendimiento con diferentes tamaños de listas:")
for tamaño, lista in listas.items():
    tiempo_burbuja = timeit.timeit(lambda: metodo_burbuja(lista.copy()), number=1)
    tiempo_quicksort = timeit.timeit(lambda: quicksort(lista.copy()), number=1)
    print(f"Tamaño {tamaño}: Burbuja: {tiempo_burbuja:.5f} s | Quicksort: {tiempo_quicksort:.5f} s")
```
